# Part 1: Local Binary Patterns (LBP)

LBP encodes texture by comparing each pixel to its 8 neighbors in a 3×3 window:

1. If neighbor ≥ center → **1**, else → **0**
2. Read the 8 bits clockwise from top-left → 8-bit binary number
3. Convert to decimal → LBP code (0 to 255)

## TODO 1: Implement LBP for a single pixel

Complete the function below to compute the LBP code for a pixel at position (row, col).

The 8 neighbors are already extracted for you in clockwise order starting from top-left (P0 to P7).

Your task: compare each neighbor to the center and convert the result to a decimal value.

**Hint:** P0 (top-left) has weight 2⁷ = 128, P7 (mid-left) has weight 2⁰ = 1.

```python
def compute_lbp_pixel(image, row, col):

    center = image[row, col]

    # 8  neighbors in clockwise order starting from top-left

    neighbors = [
        image[row - 1, col - 1],  # P0: top-left
        image[row - 1, col],      # P1: top-center
        image[row - 1, col + 1],  # P2: top-right
        image[row,     col + 1],  # P3: mid-right
        image[row + 1, col + 1],  # P4: bottom-right
        image[row + 1, col],      # P5: bottom-center
        image[row + 1, col - 1],  # P6: bottom-left
        image[row,     col - 1],  # P7: mid-left
    ]

    # TODO: Compare each neighbor to center.
    # If neighbor >= centerr, the bit is 1; otherwise 0.
    # Convert the 8 bits to a decimal number.

    lbp_code = 0
    for i, neighbor in enumerate(neighbors):
      if neighbor >= center:
        lbp_code += 2 ** (7 - i)


    return lbp_code
```

```python
# For Verification

test_patch = np.array([
    [78, 32, 90],
    [12, 55, 47],
    [61, 88, 10]
], dtype=np.uint8)

result = compute_lbp_pixel(test_patch, 1, 1)
assert result == 166, f"Expected 166, got {result}"
```

## Applying LBP to an image

Now we apply the LBP function to every pixel in the image (except border pixels) and visualize the result.

```python
def compute_lbp_image(image):

    rows, cols = image.shape
    lbp_image = np.zeros((rows, cols), dtype=np.uint8)
    for r in range(1, rows - 1):
        for c in range(1, cols - 1):
            lbp_image[r, c] = compute_lbp_pixel(image, r, c)
    return lbp_image

# Load imags

coffee_img = rgb2gray(data.coffee())
coffee_img = (coffee_img * 255).astype(np.uint8)
coffee_img = coffee_img[:256, :256]

moon_img = data.moon()
moon_img = moon_img[:256, :256]


# Compute LBP
lbp_coffee = compute_lbp_image(coffee_img)
lbp_moon = compute_lbp_image(moon_img)


# Visualize

fig, axes = plt.subplots(2, 2, figsize=(10, 8))
axes[0, 0].imshow(coffee_img, cmap='gray')
axes[0, 0].set_title('Coffee - Original')
axes[0, 0].axis('off')
axes[0, 1].imshow(moon_img, cmap='gray')
axes[0, 1].set_title('Moon - Originel')
axes[0, 1].axis('off')
axes[1, 0].imshow(lbp_coffee, cmap='gray')
axes[1, 0].set_title('Coffee - LBP')
axes[1, 0].axis('off')
axes[1, 1].imshow(lbp_moon, cmap='gray')
axes[1, 1].set_title('Moon - LBP')
axes[1, 1].axis('off')
plt.tight_layout()
plt.show()
```

## TODO 2: Build a normalized LBP histogram

Complete the function below to compute a histogram of LBP codes for an image. The histogram should be normalized so it sums to 1.0.

```python
def lbp_histogram(lbp_image, num_bins=256):
    # TODO: Use np.histogram to count occurrences of each code.
    # Set bins=num_bins and range=(0, num_bins).
    # Normalize the result so it sums to 1.0.
    hist, _ = np.histogram(lbp_image, bins=num_bins, range=(0, num_bins))
    hist = hist.astype(np.float64)
    hist /= hist.sum()

    return hist
```

```python
# For Verification

assert abs(lbp_histogram(lbp_coffee).sum() - 1.0)< 1e-6, "Histogram doesn't sum to 1"

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].bar(range(256), lbp_histogram(lbp_coffee), color='brown', width=1.0)
axes[0].set_title('Coffee - LBP Histogram')
axes[0].set_xlabel('LBP Code')
axes[1].bar(range(256), lbp_histogram(lbp_moon), color='gray', width=1.0)
axes[1].set_title('Moon - LBP Histogram')
axes[1].set_xlabel('LBP Code')
plt.tight_layout()
plt.show()
```

##TODO 1: build a GLCM for a small image

Write a function that looks at the right neighbor of each pixel (if it has one) and increments the corresponding spot in the GLCM. For this exercise, you can assume that pixel intensity ranges from 0 to 2.

```python
def create_glcm(img):
  #create empty GLCM
  glcm = np.zeros((3,3))

  #TODO: for each row:
  #for each pixel (except the last) of intensity i with a neighbor of intensity j,
  #increment GLCM[j][i]
  for row in img:
    for px in range(len(row)):
      if px != (len(row) - 1):
        glcm[row[px+1]][row[px]] += 1
  return glcm

img = np.array([
    [0, 2, 1],
    [1, 2, 1],
    [2, 0, 1],
    [1, 2, 1]
])
create_glcm(img)
```

##TODO 2: Create a baseline for textures

```python
#open the camera image
image = data.camera()
plt.imshow(image, cmap="gray")
plt.show()
```

Select 8 patches total from the image that match one of two textures: grass or sky. To select a patch, just choose the coordinate pair of its upper left corner and the provided code will create a patch of size {PATCH_SIZE}

```python

PATCH_SIZE = 21

#select 4 patches from grassy areas of the image
#hint: the image is 512x512
#eg. (342, 223), (444, 192), etc.
grass_locations = [
    #TODO
    (280, 454),
    (342, 223),
    (444, 192),
    (455, 455)
]
grass_patches = []
for loc in grass_locations:
    grass_patches.append(
        image[loc[0] : loc[0] + PATCH_SIZE, loc[1] : loc[1] + PATCH_SIZE]
    )

#select 4 patches from sky areas of the image
#eg. (139, 28), (145, 379), etc.
sky_locations = [
    #TODO
    (38, 34),
    (139, 28),
    (37, 437),
    (145, 379)
]
sky_patches = []
for loc in sky_locations:
    sky_patches.append(
        image[loc[0] : loc[0] + PATCH_SIZE, loc[1] : loc[1] + PATCH_SIZE]
    )
#create the figure
fig = plt.figure(figsize=(8, 8))

#display original image with locations of patches
ax = fig.add_subplot(3, 1, 1)
ax.imshow(image, cmap=plt.cm.gray, vmin=0, vmax=255)
for y, x in grass_locations:
    ax.plot(x + PATCH_SIZE / 2, y + PATCH_SIZE / 2, 'gs', label="Grass")
for y, x in sky_locations:
    ax.plot(x + PATCH_SIZE / 2, y + PATCH_SIZE / 2, 'bs', label="Sky")
ax.set_xlabel('Original Image')
ax.set_xticks([])
ax.set_yticks([])
ax.axis('image')
#ax.legend()

#display the image patches
for i, patch in enumerate(grass_patches):
    ax = fig.add_subplot(3, len(grass_patches), len(grass_patches) * 1 + i + 1)
    ax.imshow(patch, cmap=plt.cm.gray, vmin=0, vmax=255)
    ax.set_xlabel(f"Grass {i + 1}")

for i, patch in enumerate(sky_patches):
    ax = fig.add_subplot(3, len(sky_patches), len(sky_patches) * 2 + i + 1)
    ax.imshow(patch, cmap=plt.cm.gray, vmin=0, vmax=255)
    ax.set_xlabel(f"Sky {i + 1}")

plt.tight_layout()
plt.show()
```

We will then create the GLCM and use features (homogeneity and correlation) calculated from it to analyze the patches.

```python

#compute GLCM features
xs = []
ys = []
for patch in grass_patches + sky_patches:
    glcm = graycomatrix(
        patch, distances=[5], angles=[0], levels=256, symmetric=True, normed=True
    )
    xs.append(graycoprops(glcm, 'homogeneity')[0, 0])
    ys.append(graycoprops(glcm, 'correlation')[0, 0])

#create the figure
fig = plt.figure(figsize=(8, 8))

#for each patch, plot (homogeneity, correlation)
ax = fig.add_subplot(3, 2, 2)
ax.plot(xs[: len(grass_patches)], ys[: len(grass_patches)], 'go', label='Grass')
ax.plot(xs[len(grass_patches) :], ys[len(grass_patches) :], 'bo', label='Sky')
ax.set_xlabel('GLCM Homogeneity')
ax.set_ylabel('GLCM Correlation')
ax.legend()
fig.suptitle('Grey level co-occurrence matrix features', fontsize=14, y=1.05)
plt.tight_layout()
plt.show()
```
